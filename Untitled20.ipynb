```python
numeric_cols = [
    "completion_rate",
    "avg_score_improvement",
    "avg_quiz_score",
    "dropout_rate",
    "avg_watch_time",
    "assignment_submission_rate",
    "forum_activity_rate",
    "avg_feedback_score",
    "feedback_response_rate"
]
```

```python
for col in numeric_cols:
    if col in data.columns:
        converted = pd.to_numeric(data[col], errors="coerce")
        invalid = data[col].notna() & converted.isna()

        print(f"\n{col}:")
        print("Invalid values:", data.loc[invalid, col].unique())
```

```python
data.head(10)
```

```python
#Rate columns check
rate_cols = [
    "completion_rate",
    "dropout_rate",
    "assignment_submission_rate",
    "forum_activity_rate",
    "feedback_response_rate"
]

for col in rate_cols:
    if col in data.columns:
        print(f"\n{col}")
        print("Min:", data[col].min())
        print("Max:", data[col].max())
```

```python
# Check negative values in numeric columns
num_data = data.select_dtypes(include="number")

for col in num_data.columns:
    negative_count = (data[col] < 0).sum()
    if negative_count > 0:
        print(col, "has", negative_count, "negative values")
```

```python
num_data
```

```python
# Step 2: Range validation

rate_cols = [
    "completion_rate",
    "dropout_rate",
    "assignment_submission_rate",
    "forum_activity_rate",
    "feedback_response_rate"
]

for col in rate_cols:
    if col in data.columns:
        invalid = data[~data[col].between(0, 1)]

        print(f"\n{col}")
        print("Invalid rows:", len(invalid))

        if not invalid.empty:
            display(invalid[[col]].head())
```

```python
# Check minimum and maximum values
check_cols = [
    "avg_quiz_score",
    "avg_feedback_score",
    "avg_watch_time",
    "avg_score_improvement"
]

for col in check_cols:
    if col in data.columns:
        print(f"\n{col}")
        print("Minimum:", data[col].min())
        print("Maximum:", data[col].max())
```

```python
# Check missing values and duplicates again
print("Total missing values:", data.isnull().sum().sum())
print("Total duplicate rows:", data.duplicated().sum())

# Check negative values
num_data = data.select_dtypes(include="number")
print("\nNegative values per column:")
display((num_data < 0).sum().to_frame(name="negative_count"))

# Check whether rate columns stay between 0 and 1
rate_cols = [
    "completion_rate",
    "dropout_rate",
    "assignment_submission_rate",
    "forum_activity_rate",
    "feedback_response_rate"
]

for col in rate_cols:
    if col in data.columns:
        print(
            col,
            "out-of-range rows:",
            (~data[col].between(0, 1)).sum()
        )
```

```python
# Check data types of all columns
print("Column Data Types:")
display(data.dtypes.to_frame(name="data_type"))

# Check unique values in ID columns
id_cols = ["batch_id", "instructor_id", "course_id"]

for col in id_cols:
    if col in data.columns:
        print(f"{col}: {data[col].nunique()} unique values")
```

# Exploratory Data Analysis karenge.- EDA

```python
# Univariant analysis
# Calculate mean and median for numerical columns
numeric_cols = data.select_dtypes(include="number").columns

summary = pd.DataFrame({
    "Mean": data[numeric_cols].mean(),
    "Median": data[numeric_cols].median(),
    "Std_Deviation": data[numeric_cols].std(),
    "Minimum": data[numeric_cols].min(),
    "Maximum": data[numeric_cols].max()
})

display(summary.round(2))
```

```python
# histogram
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.histplot(data=data, x="completion_rate", bins=20, kde=True)

plt.title("Distribution of Course Completion Rate")
plt.xlabel("Completion Rate")
plt.ylabel("Frequency")

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

numeric_cols = data.select_dtypes(include="number").columns

fig, axes = plt.subplots(
    nrows=len(numeric_cols),
    ncols=2,
    figsize=(14, 4 * len(numeric_cols))
)

for i, col in enumerate(numeric_cols):

    # Histogram
    sns.histplot(
        data=data,
        x=col,
        bins=20,
        kde=True,
        ax=axes[i, 0]
    )
    axes[i, 0].set_title(f"Distribution of {col}")

    # Box Plot
    sns.boxplot(
        data=data,
        x=col,
        ax=axes[i, 1]
    )
    axes[i, 1].set_title(f"Box Plot of {col}")

plt.tight_layout()
plt.show()
```
