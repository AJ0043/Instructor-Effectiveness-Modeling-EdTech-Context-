```python

# Identify potential outliers using the IQR method

outlier_results = []
for col in numeric_cols:
    Q1 = data[col].quantile(0.25)
    Q3 = data[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = data[
        (data[col] < lower_bound) |
        (data[col] > upper_bound)
    ]

    outlier_results.append({
        "Feature": col,
        "Lower Bound": round(lower_bound, 3),
        "Upper Bound": round(upper_bound, 3),
        "Outlier Count": len(outliers),
        "Outlier Percentage": round(
            len(outliers) / len(data) * 100, 2
        )
    })

outlier_summary = pd.DataFrame(outlier_results)

display(outlier_summary)
```

Step 2:- Instructor Effectiveness Score

```python

from sklearn.preprocessing import MinMaxScaler

# Metrics used to calculate instructor effectiveness
positive_metrics = [
    "completion_rate",
    "avg_score_improvement",
    "avg_quiz_score",
    "assignment_submission_rate",
    "avg_feedback_score"
]

# Normalize positive metrics to a 0–1 scale
scaler = MinMaxScaler()
data_scaled = data.copy()

data_scaled[positive_metrics] = scaler.fit_transform(
    data[positive_metrics]
)

# Lower dropout is better, so reverse its normalized value
data_scaled["dropout_effectiveness"] = (
    1 - MinMaxScaler().fit_transform(data[["dropout_rate"]]).ravel()
)

# Calculate a weighted effectiveness score (0–100)
data_scaled["instructor_effectiveness_score"] = 100 * (
    0.25 * data_scaled["completion_rate"] +
    0.20 * data_scaled["avg_score_improvement"] +
    0.15 * data_scaled["avg_quiz_score"] +
    0.15 * data_scaled["dropout_effectiveness"] +
    0.15 * data_scaled["assignment_submission_rate"] +
    0.10 * data_scaled["avg_feedback_score"]
)

display(
    data_scaled[
        ["instructor_id", "instructor_effectiveness_score"]
    ].head()
)
```

### Batch-Level Instructor Effectiveness Score

In this step, a weighted instructor effectiveness score was calculated for each batch record using key educational performance indicators. Min-Max Scaling was applied to normalize the positive performance metrics to a range of 0 to 1. The dropout rate was inversely scaled because a lower dropout rate represents better performance. Six indicators were assigned different weights: completion rate (25%), average score improvement (20%), average quiz score (15%), dropout effectiveness (15%), assignment submission rate (15%), and average feedback score (10%). These weighted metrics were combined to calculate an effectiveness score ranging from 0 to 100. The resulting scores provide an initial basis for classifying performance into Low, Medium, and High effectiveness tiers.


```python
# Step 3 define Effectiveness tier

data_scaled["effectiveness_tier"] = pd.cut(
    data_scaled["instructor_effectiveness_score"],
    bins=[-float("inf"), 40, 70, float("inf")],
    labels=["Low", "Medium", "High"]
)

display(
    data_scaled[
        ["instructor_id", "instructor_effectiveness_score", "effectiveness_tier"]
    ].head(10)
)

print("\nTier Distribution:")
print(data_scaled["effectiveness_tier"].value_counts())
```

### Definition of Instructor Effectiveness Tiers

In this step, instructor effectiveness scores were categorized into three tiers: Low, Medium, and High using the Pandas `cut()` function. Scores up to 40 were classified as Low, scores above 40 and up to 70 as Medium, and scores above 70 as High. This classification converts continuous effectiveness scores into meaningful categories, making it easier to compare performance levels and identify instructors who may require additional support or whose teaching practices could be used as examples of effective instruction. The tier distribution was also examined to understand how the records were distributed across the three categories.


```python
print(
    data_scaled["effectiveness_tier"]
    .value_counts()
    .reindex(["Low", "Medium", "High"], fill_value=0)
)
```

```python
# Step 3- Aggregate batch data to instructor level

instructor_data = data_scaled.groupby("instructor_id").agg(
    total_batches=("batch_id", "nunique"),
    completion_rate=("completion_rate", "mean"),
    avg_score_improvement=("avg_score_improvement", "mean"),
    avg_quiz_score=("avg_quiz_score", "mean"),
    dropout_rate=("dropout_rate", "mean"),
    avg_watch_time=("avg_watch_time", "mean"),
    assignment_submission_rate=("assignment_submission_rate", "mean"),
    forum_activity_rate=("forum_activity_rate", "mean"),
    avg_feedback_score=("avg_feedback_score", "mean"),
    feedback_response_rate=("feedback_response_rate", "mean")
).reset_index()

display(instructor_data.head(10))

print("Total instructors:", instructor_data["instructor_id"].nunique())
print("Total aggregated rows:", len(instructor_data))
print("Minimum batches per instructor:", instructor_data["total_batches"].min())
print("Maximum batches per instructor:", instructor_data["total_batches"].max())
```

### Aggregation of Batch Data at Instructor Level

In this step, batch-level data was aggregated to the instructor level using the `groupby()` and `agg()` functions in Pandas. Each instructor's performance metrics were summarized by calculating their mean values across all associated batches. The number of unique batches handled by each instructor was also calculated using `nunique()`. This process transformed the dataset into an instructor-level dataset, allowing overall performance to be compared across instructors. The aggregation produced 120 instructor records, with each instructor handling between 7 and 31 batches.


```python
#Calculate instructor effectiveness score#

from sklearn.preprocessing import MinMaxScaler

instructor_data["effectiveness_score"] = 0.0

positive_metrics = [
    "completion_rate",
    "avg_score_improvement",
    "avg_quiz_score",
    "assignment_submission_rate",
    "avg_feedback_score"
]

scaler = MinMaxScaler()
instructor_data[positive_metrics] = scaler.fit_transform(
    instructor_data[positive_metrics]
)

instructor_data["dropout_effectiveness"] = (
    1 - MinMaxScaler().fit_transform(
        instructor_data[["dropout_rate"]]
    ).ravel()
)

instructor_data["effectiveness_score"] = 100 * (
    0.25 * instructor_data["completion_rate"] +
    0.20 * instructor_data["avg_score_improvement"] +
    0.15 * instructor_data["avg_quiz_score"] +
    0.15 * instructor_data["dropout_effectiveness"] +
    0.15 * instructor_data["assignment_submission_rate"] +
    0.10 * instructor_data["avg_feedback_score"]
)

display(
    instructor_data[
        ["instructor_id", "total_batches", "effectiveness_score"]
    ].head(10)
)
```

Instructor Effectiveness Score Calculation

In this step, a composite instructor effectiveness score was calculated using the aggregated instructor-level data. Min-Max Scaling was applied to normalize the selected performance metrics to a common scale. The score was calculated using six key indicators: completion rate (25%), average score improvement (20%), average quiz score (15%), dropout effectiveness (15%), assignment submission rate (15%), and average feedback score (10%). Dropout rate was inversely scaled because a lower dropout rate indicates better performance. The weighted scores were combined to produce a final effectiveness score ranging from 0 to 100, allowing instructors to be compared based on their overall performance.

```python
# Calculate Instructor-Level Effectiveness Score Using Weighted Metrics #

instructor_scored = instructor_data.copy()

positive_metrics = [
    "completion_rate",
    "avg_score_improvement",
    "avg_quiz_score",
    "assignment_submission_rate",
    "avg_feedback_score"
]

scaled_metrics = MinMaxScaler().fit_transform(
    instructor_scored[positive_metrics]
)

for i, col in enumerate(positive_metrics):
    instructor_scored[col + "_scaled"] = scaled_metrics[:, i]

instructor_scored["dropout_effectiveness"] = (
    1 - MinMaxScaler().fit_transform(
        instructor_scored[["dropout_rate"]]
    ).ravel()
)

instructor_scored["effectiveness_score"] = 100 * (
    0.25 * instructor_scored["completion_rate_scaled"] +
    0.20 * instructor_scored["avg_score_improvement_scaled"] +
    0.15 * instructor_scored["avg_quiz_score_scaled"] +
    0.15 * instructor_scored["dropout_effectiveness"] +
    0.15 * instructor_scored["assignment_submission_rate_scaled"] +
    0.10 * instructor_scored["avg_feedback_score_scaled"]
)

display(
    instructor_scored[
        ["instructor_id", "total_batches", "effectiveness_score"]
    ].head(10)
)
```

### Instructor-Level Effectiveness Score Calculation

In this step, an overall effectiveness score was calculated for each instructor using aggregated batch-level data. Min-Max Scaling was applied to normalize five positive performance metrics: completion rate, average score improvement, average quiz score, assignment submission rate, and average feedback score. Dropout rate was inversely scaled because a lower dropout rate indicates better performance. A weighted scoring approach was used, assigning 25% to completion rate, 20% to score improvement, 15% each to quiz score, dropout effectiveness, and assignment submission rate, and 10% to feedback score. The final score was calculated on a scale of 0 to 100 to support comparison of instructor performance.


```python
# Define instructor-level effectiveness tiers

instructor_scored["effectiveness_tier"] = pd.cut(
    instructor_scored["effectiveness_score"],
    bins=[-float("inf"), 40, 70, float("inf")],
    labels=["Low", "Medium", "High"]
)

display(
    instructor_scored[
        ["instructor_id", "total_batches",
         "effectiveness_score", "effectiveness_tier"]
    ].head(10)
)

print("\nInstructor Effectiveness Tier Distribution:")
print(
    instructor_scored["effectiveness_tier"]
    .value_counts()
    .reindex(["Low", "Medium", "High"], fill_value=0)
)
```

### Instructor Effectiveness Tier Classification

In this step, instructors were classified into three effectiveness categories—Low, Medium, and High—based on their calculated effectiveness scores. The Pandas `cut()` function was used to assign tiers using predefined thresholds: scores up to 40 were classified as Low, scores above 40 and up to 70 as Medium, and scores above 70 as High. The classification results were displayed alongside instructor IDs, total batches handled, and effectiveness scores. Finally, the number of instructors in each tier was calculated to understand the distribution of overall instructor effectiveness.


```python

# Step 4 - Prepare data for Machine Learning

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

features = [
    "completion_rate",
    "avg_score_improvement",
    "avg_quiz_score",
    "dropout_rate",
    "avg_watch_time",
    "assignment_submission_rate",
    "forum_activity_rate",
    "avg_feedback_score",
    "feedback_response_rate",
    "total_batches"
]

X = instructor_data[features]

y = instructor_scored["effectiveness_tier"]

print("Feature dataset shape:", X.shape)
print("Target dataset shape:", y.shape)
print("\nTarget distribution:")
print(y.value_counts())
```

### Preparation of Data for Machine Learning

In this step, the instructor-level dataset was prepared for classification using a Random Forest model. The performance metrics, including completion rate, score improvement, quiz score, dropout rate, watch time, assignment submission, forum activity, feedback score, feedback response rate, and total batches handled, were selected as input features (`X`). The instructor effectiveness tier—Low, Medium, or High—was selected as the target variable (`y`). The shapes of the feature and target datasets and the target class distribution were examined to verify the data before model training.


```python

# Step 4.2 - Split data into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training features shape:", X_train.shape)
print("Testing features shape:", X_test.shape)
print("Training target shape:", y_train.shape)
print("Testing target shape:", y_test.shape)
```

### Splitting Data into Training and Testing Sets

In this step, the instructor-level dataset was divided into training and testing sets using Scikit-learn's `train_test_split()` function. A split ratio of 80% for training and 20% for testing was applied. The `random_state=42` parameter was used to ensure reproducibility, while `stratify=y` maintained approximately the same proportion of Low, Medium, and High effectiveness tiers in both sets. The training dataset will be used to train the Random Forest classifier, while the testing dataset will be used to evaluate its classification performance on unseen records.


```python

# Step 4.3 - Train the Random Forest Classifier

rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    class_weight="balanced"
)

rf_model.fit(X_train, y_train)

print("Random Forest model trained successfully!")
```

### Training the Random Forest Classifier

A Random Forest classification model was developed using Scikit-learn to classify instructors into Low, Medium, and High effectiveness tiers. The model was configured with 100 decision trees and a fixed random state to ensure reproducibility. Balanced class weights were used to account for potential differences in the number of records across effectiveness categories. The model was trained using the training dataset and prepared for evaluation on the held-out test dataset.


```python

# Step 4.4 - Predict Instructor Effectiveness Tiers

y_pred = rf_model.predict(X_test)

print("Actual effectiveness tiers:")
print(y_test.head(10).to_list())

print("\nPredicted effectiveness tiers:")
print(y_pred[:10])
```

### Prediction of Instructor Effectiveness Tiers

After training the Random Forest classifier, predictions were generated using the testing dataset. The predicted effectiveness tiers were compared with the actual tier labels to assess the model's classification performance. These predictions were subsequently used to calculate evaluation metrics and examine classification errors.


```python

# Step 5.1 - Evaluate the Random Forest Model

accuracy = accuracy_score(y_test, y_pred)

print("Model Accuracy:", round(accuracy * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))

print("\nConfusion Matrix:")
print(confusion_matrix(
    y_test,
    y_pred,
    labels=["Low", "Medium", "High"]
))
```

### Evaluation of the Random Forest Classifier

The trained Random Forest classifier was evaluated using the held-out test dataset and achieved an overall accuracy of **87.5%**. Accuracy was used to measure the proportion of correct predictions. Precision, recall, and F1-score were examined for each effectiveness tier to assess the model's classification performance across categories. A confusion matrix was also generated to identify correct classifications and misclassifications among Low, Medium, and High effectiveness tiers.


```python

# Step 5.2 - Visualize Confusion Matrix

import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(
    y_test,
    y_pred,
    labels=["Low", "Medium", "High"]
)

plt.figure(figsize=(7, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Greens",
    xticklabels=["Low", "Medium", "High"],
    yticklabels=["Low", "Medium", "High"]
)

plt.title("Confusion Matrix - Random Forest Classifier")
plt.xlabel("Predicted Tier")
plt.ylabel("Actual Tier")
plt.tight_layout()
plt.show()
```
